# Flash and test the Barrett firmware

Run `make` in this folder first. Then run the cells in order.

```python
import chipwhisperer as cw

# Connect to the Husky and the SAM4S target
scope = cw.scope()
target = cw.target(scope)
scope.default_setup()
```

```python
# Write the firmware to the SAM4S
cw.program_target(scope, cw.programmers.SAM4SProgrammer, "simpleserial-barrett-CWHUSKY.hex")
```

Send `x` with the `b` command and read `x mod N` back with `r`. `N` must match the value in `simpleserial-barrett.c`.

```python
N = 17669
x = 1212218

target.simpleserial_write('b', bytearray(x.to_bytes(3, 'little')))
r = int.from_bytes(target.simpleserial_read('r', 4), 'little')

print(r, x % N)
assert r == x % N
```

```python
# Release the Husky for other programs
scope.dis()
target.dis()
```
